```python
from scipy.ndimage import gaussian_filter1d

def smooth_with_gaussian(x, y, sigma=1):
    """
    Apply Gaussian kernel smoothing to the data.
    
    Parameters:
    - x: Array of x-values.
    - y: Array of y-values.
    - sigma: The standard deviation of the Gaussian kernel.
    
    Returns:
    - y_smooth: The smoothed y-values.
    """
    y_smooth = gaussian_filter1d(y, sigma=sigma)
    return x, y_smooth

DM_Linden_x, DM_Linden_y = smooth_with_gaussian(DM_Linden_x, DM_Linden_y)
DM_coogan_x, DM_coogan_y = smooth_with_gaussian(DM_coogan_x, DM_coogan_y)
DM_Korsmeier_upper_x, DM_Korsmeier_upper_y = smooth_with_gaussian(DM_Korsmeier_upper_x, DM_Korsmeier_upper_y)
DM_Korsmeier_lower_x, DM_Korsmeier_lower_y = smooth_with_gaussian(DM_Korsmeier_lower_x, DM_Korsmeier_lower_y)
DM_Nan_Li_upper_x, DM_Nan_Li_upper_y = smooth_with_gaussian(DM_Nan_Li_upper_x, DM_Nan_Li_upper_y)
DM_Nan_Li_lower_x, DM_Nan_Li_lower_y = smooth_with_gaussian(DM_Nan_Li_lower_x, DM_Nan_Li_lower_y)
DM_Nan_Li_x, DM_Nan_Li_y = smooth_with_gaussian(DM_Nan_Li_x, DM_Nan_Li_y)
DM_MKachelrieβ_x, DM_MKachelrieβ_y = smooth_with_gaussian(DM_MKachelrieβ_x, DM_MKachelrieβ_y)
DM_M_Winkler_x, DM_M_Winkler_y = smooth_with_gaussian(DM_M_Winkler_x, DM_M_Winkler_y)
BKG_Blum_x, BKG_Blum_y = smooth_with_gaussian(BKG_Blum_x, BKG_Blum_y)
BKG_Poulin_x, BKG_Poulin_y = smooth_with_gaussian(BKG_Poulin_x, BKG_Poulin_y)
BKG_A_Shukla_x, BKG_A_Shukla_y = smooth_with_gaussian(BKG_A_Shukla_x, BKG_A_Shukla_y)
BKG_M_Kachelrie_background_x, BKG_M_Kachelrie_background_y = smooth_with_gaussian(BKG_M_Kachelrie_background_x, BKG_M_Kachelrie_background_y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Data
log_array = [1, 0.999, 0.994, 0.982, 0.885, 0.959, 0.988, 0.986, 0.969, 0.999, 1, 0.997, 1, 1, 1, 1]
stick = ["5-6 ns", "6-7 ns", "7-8 ns", "8-9 ns", "9-10 ns", "10-11 ns", "11-12 ns", "12-13 ns", "13-14 ns", "14-15 ns", "15-16 ns", "16-17 ns", "17-18 ns", "18-19 ns", "19-20 ns", "20-21 ns"]

# Function to format y-axis labels as percentages
def to_percent(y, _):
    #return f"{y * 100:.0f}%"
    return f"{y:.02f}"

# Create a figure with two subplots (one for the broken axis effect)
fig, (ax2, ax1) = plt.subplots(2, 1, sharex=True, gridspec_kw={'height_ratios': [5, 1]}, figsize=(6, 6))

# Plot the data on both subplots
ax1.plot(log_array, marker='.', linestyle='-', color='b')
ax2.plot(log_array, marker='.', linestyle='-', color='b', label='Antihelium-3 selection efficiency')

# Set y-axis limits for the broken axis effect
ax1.set_ylim(0, 0.05)  # Lower part of the broken axis (0% to 20%)
ax2.set_ylim(0.85, 1.01)  # Upper part of the broken axis (80% to 100%)

# Hide the spines between the two subplots
ax1.spines['top'].set_visible(False)
ax2.spines['bottom'].set_visible(False)
ax1.xaxis.tick_bottom()
ax2.xaxis.tick_top()
ax2.tick_params(labeltop=False)  # Hide x-axis labels for the top subplot

# Add diagonal lines to indicate the broken axis
d = 0.5  # Size of the diagonal lines
kwargs = dict(marker=[(-1, -d), (1, d)], markersize=12, linestyle="none", color='k', mec='k', mew=1, clip_on=False)
ax1.plot([0, 1], [1, 1], transform=ax1.transAxes, **kwargs)
ax2.plot([0, 1], [0, 0], transform=ax2.transAxes, **kwargs)

# Format y-axis labels as percentages
ax1.yaxis.set_major_formatter(FuncFormatter(to_percent))
ax2.yaxis.set_major_formatter(FuncFormatter(to_percent))

# Add stick labels to x-axis
ax1.set_xticks(range(len(stick)))
ax1.set_xticklabels(stick, rotation=45)

# Enable finer grid
ax1.grid(True, which="both", ls="--", linewidth=0.5)
ax2.grid(True, which="both", ls="--", linewidth=0.5)

# Labels and title
ax1.set_xlabel('Time in between TOF paddles (ns)', fontsize=16)
#ax1.set_ylabel('Percentage (0-20%)', fontsize=16)
ax2.set_ylabel('Selection efficiency', fontsize=16)

# Adjust tick font size
ax1.tick_params(axis='both', labelsize=10)
ax2.tick_params(axis='both', labelsize=10)

# Show legend
ax2.legend(fontsize=10, loc="right")

# Adjust layout to prevent label overlap
plt.tight_layout()

# Show plot
plt.show()
```
